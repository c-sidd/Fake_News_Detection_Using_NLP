## 2. Text Preprocessing and Cleaning

In this section, we clean and preprocess the news headline text (`title`):
- Convert text to lowercase
- Remove special characters, symbols, and punctuation while preserving unicode word characters
- Normalize excessive whitespace
- Verify missing or empty values after cleaning

```python
import re

def clean_text(text):
    if not isinstance(text, str):
        return ""
    # Convert to lowercase
    text = text.lower()
    # Remove special characters and punctuation, preserving words and spaces
    text = re.sub(r'[^\w\s]', '', text, flags=re.UNICODE)
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Demonstration of text cleaning on a sample headline
sample_raw = X_train['title'].iloc[5]
print("--- Raw Headline ---")
print(sample_raw)
print("\n--- Cleaned Headline ---")
print(clean_text(sample_raw))
```

```python
# Apply cleaning to both training and testing datasets
X_train = X_train.copy()
X_test = X_test.copy()

X_train['cleaned_title'] = X_train['title'].apply(clean_text)
X_test['cleaned_title'] = X_test['title'].apply(clean_text)

print("Cleaned titles preview:")
display(X_train[['title', 'cleaned_title']].head())

print("\nEmpty cleaned titles in Train:", (X_train['cleaned_title'].str.len() == 0).sum())
print("Empty cleaned titles in Test :", (X_test['cleaned_title'].str.len() == 0).sum())
```

```python
# Verify class distribution in Train and Test sets
print("Training Class Distribution:")
print(y_train.value_counts())
print("\nTraining Class Ratio (%):")
print((y_train.value_counts(normalize=True) * 100).round(2))

print("\nTesting Class Distribution:")
print(y_test.value_counts())
print("\nTesting Class Ratio (%):")
print((y_test.value_counts(normalize=True) * 100).round(2))

plt.figure(figsize=(6, 4))
sns.countplot(x=y_train, hue=y_train, palette='viridis', legend=False)
plt.title("Label Distribution in Training Set (0: Fake, 1: Real)")
plt.xlabel("Label")
plt.ylabel("Count")
plt.show()
```

## 3. Feature Extraction (TF-IDF Vectorization)

We use **Term Frequency - Inverse Document Frequency (TF-IDF)** to convert text into numerical features:
1. **Unigram TF-IDF**: Captures individual word frequencies (up to 10,000 features).
2. **Bigram TF-IDF**: Captures word pairs (`ngram_range=(1, 2)`, up to 15,000 features) to retain contextual combinations.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Unigram TF-IDF Vectorizer
tfidf = TfidfVectorizer(
    max_features=10000,
    stop_words='english'
)

X_train_tfidf = tfidf.fit_transform(X_train['cleaned_title'])
X_test_tfidf = tfidf.transform(X_test['cleaned_title'])

print("Unigram TF-IDF - Train shape:", X_train_tfidf.shape)
print("Unigram TF-IDF - Test shape :", X_test_tfidf.shape)
```

```python
# 2. Bigram TF-IDF Vectorizer (Unigrams + Bigrams)
tfidf_bigram = TfidfVectorizer(
    ngram_range=(1, 2),
    max_features=15000,
    stop_words='english'
)

X_train_tfidf_bigram = tfidf_bigram.fit_transform(X_train['cleaned_title'])
X_test_tfidf_bigram = tfidf_bigram.transform(X_test['cleaned_title'])

print("Bigram TF-IDF - Train shape:", X_train_tfidf_bigram.shape)
print("Bigram TF-IDF - Test shape :", X_test_tfidf_bigram.shape)
```

## 4. Model Training & Comparison

We train and compare multiple machine learning algorithms:
1. **Logistic Regression (Baseline)**
2. **Logistic Regression (Class Weighted / Balanced)**
3. **Multinomial Naive Bayes**
4. **Linear Support Vector Classifier (LinearSVC with Bigrams)**
5. **Hyperparameter Tuned Logistic Regression (GridSearchCV)**

```python
# Initialize list to track performance across all models
model_results = []
```

```python
# Model 1: Logistic Regression (Baseline)
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix

lr_baseline = LogisticRegression(max_iter=1000, random_state=42)
lr_baseline.fit(X_train_tfidf, y_train)

y_pred_baseline = lr_baseline.predict(X_test_tfidf)

acc_baseline = accuracy_score(y_test, y_pred_baseline)
prec_baseline = precision_score(y_test, y_pred_baseline)
rec_baseline = recall_score(y_test, y_pred_baseline)
f1_baseline = f1_score(y_test, y_pred_baseline)

print("--- Model 1: Logistic Regression (Baseline) ---")
print(f"Accuracy : {acc_baseline:.4f}")
print(f"Precision: {prec_baseline:.4f}")
print(f"Recall   : {rec_baseline:.4f}")
print(f"F1 Score : {f1_baseline:.4f}\n")
print(classification_report(y_test, y_pred_baseline, target_names=['Fake News (0)', 'Real News (1)']))

model_results.append({
    'Model': 'Logistic Regression (Baseline)',
    'Accuracy': acc_baseline,
    'Precision': prec_baseline,
    'Recall': rec_baseline,
    'F1 Score': f1_baseline
})
```

```python
# Model 2: Logistic Regression (Balanced Class Weight)
lr_balanced = LogisticRegression(
    class_weight='balanced',
    max_iter=1000,
    random_state=42
)
lr_balanced.fit(X_train_tfidf, y_train)

y_pred_balanced = lr_balanced.predict(X_test_tfidf)

acc_balanced = accuracy_score(y_test, y_pred_balanced)
prec_balanced = precision_score(y_test, y_pred_balanced)
rec_balanced = recall_score(y_test, y_pred_balanced)
f1_balanced = f1_score(y_test, y_pred_balanced)

print("--- Model 2: Logistic Regression (Balanced) ---")
print(f"Accuracy : {acc_balanced:.4f}")
print(f"Precision: {prec_balanced:.4f}")
print(f"Recall   : {rec_balanced:.4f}")
print(f"F1 Score : {f1_balanced:.4f}\n")
print(classification_report(y_test, y_pred_balanced, target_names=['Fake News (0)', 'Real News (1)']))

model_results.append({
    'Model': 'Logistic Regression (Balanced)',
    'Accuracy': acc_balanced,
    'Precision': prec_balanced,
    'Recall': rec_balanced,
    'F1 Score': f1_balanced
})
```

```python
# Model 3: Multinomial Naive Bayes
from sklearn.naive_bayes import MultinomialNB

nb_model = MultinomialNB()
nb_model.fit(X_train_tfidf, y_train)

y_pred_nb = nb_model.predict(X_test_tfidf)

acc_nb = accuracy_score(y_test, y_pred_nb)
prec_nb = precision_score(y_test, y_pred_nb)
rec_nb = recall_score(y_test, y_pred_nb)
f1_nb = f1_score(y_test, y_pred_nb)

print("--- Model 3: Multinomial Naive Bayes ---")
print(f"Accuracy : {acc_nb:.4f}")
print(f"Precision: {prec_nb:.4f}")
print(f"Recall   : {rec_nb:.4f}")
print(f"F1 Score : {f1_nb:.4f}\n")
print(classification_report(y_test, y_pred_nb, target_names=['Fake News (0)', 'Real News (1)']))

model_results.append({
    'Model': 'Multinomial Naive Bayes',
    'Accuracy': acc_nb,
    'Precision': prec_nb,
    'Recall': rec_nb,
    'F1 Score': f1_nb
})
```

```python
# Model 4: Linear Support Vector Classifier (LinearSVC with Bigrams)
from sklearn.svm import LinearSVC

svm_model = LinearSVC(random_state=42)
svm_model.fit(X_train_tfidf_bigram, y_train)

y_pred_svm = svm_model.predict(X_test_tfidf_bigram)

acc_svm = accuracy_score(y_test, y_pred_svm)
prec_svm = precision_score(y_test, y_pred_svm)
rec_svm = recall_score(y_test, y_pred_svm)
f1_svm = f1_score(y_test, y_pred_svm)

print("--- Model 4: LinearSVC (with Bigrams) ---")
print(f"Accuracy : {acc_svm:.4f}")
print(f"Precision: {prec_svm:.4f}")
print(f"Recall   : {rec_svm:.4f}")
print(f"F1 Score : {f1_svm:.4f}\n")
print(classification_report(y_test, y_pred_svm, target_names=['Fake News (0)', 'Real News (1)']))

model_results.append({
    'Model': 'LinearSVC (Bigrams)',
    'Accuracy': acc_svm,
    'Precision': prec_svm,
    'Recall': rec_svm,
    'F1 Score': f1_svm
})
```

```python
# Model 5: Hyperparameter Tuning with GridSearchCV
from sklearn.model_selection import GridSearchCV

param_grid = {
    'C': [0.1, 0.5, 1, 2, 5]
}

grid_lr = LogisticRegression(class_weight='balanced', max_iter=1000, random_state=42)

grid_search = GridSearchCV(
    estimator=grid_lr,
    param_grid=param_grid,
    scoring='f1',
    cv=3,
    n_jobs=-1
)

grid_search.fit(X_train_tfidf, y_train)

print("Best Parameters:", grid_search.best_params_)
print(f"Best Cross-Validation F1: {grid_search.best_score_:.4f}")

tuned_model = grid_search.best_estimator_
y_pred_tuned = tuned_model.predict(X_test_tfidf)

acc_tuned = accuracy_score(y_test, y_pred_tuned)
prec_tuned = precision_score(y_test, y_pred_tuned)
rec_tuned = recall_score(y_test, y_pred_tuned)
f1_tuned = f1_score(y_test, y_pred_tuned)

print("\n--- Model 5: Tuned Logistic Regression ---")
print(f"Accuracy : {acc_tuned:.4f}")
print(f"Precision: {prec_tuned:.4f}")
print(f"Recall   : {rec_tuned:.4f}")
print(f"F1 Score : {f1_tuned:.4f}\n")
print(classification_report(y_test, y_pred_tuned, target_names=['Fake News (0)', 'Real News (1)']))

model_results.append({
    'Model': 'Tuned Logistic Regression (GridSearch)',
    'Accuracy': acc_tuned,
    'Precision': prec_tuned,
    'Recall': rec_tuned,
    'F1 Score': f1_tuned
})
```

## 6. Error Analysis

Analyzing classification errors reveals model behavior on tricky inputs:
- **False Positives**: Fake News misclassified as Real.
- **False Negatives**: Real News misclassified as Fake.

```python
error_df = X_test.copy()
error_df['actual'] = y_test
error_df['predicted'] = y_pred_tuned
error_df['is_correct'] = (error_df['actual'] == error_df['predicted'])

print(f"Total Test Samples    : {len(error_df)}")
print(f"Correct Predictions   : {error_df['is_correct'].sum()} ({error_df['is_correct'].mean():.2%})")
print(f"Incorrect Predictions : {(~error_df['is_correct']).sum()} ({(~error_df['is_correct']).mean():.2%})")

false_positives = error_df[(error_df['actual'] == 0) & (error_df['predicted'] == 1)]
false_negatives = error_df[(error_df['actual'] == 1) & (error_df['predicted'] == 0)]

print(f"\nFalse Positives (Fake predicted as Real): {len(false_positives)}")
print(f"False Negatives (Real predicted as Fake): {len(false_negatives)}")

print("\n--- Top 5 False Positives (Fake News Classified as Real) ---")
display(false_positives[['title', 'source_domain', 'actual', 'predicted']].head(5))

print("\n--- Top 5 False Negatives (Real News Classified as Fake) ---")
display(false_negatives[['title', 'source_domain', 'actual', 'predicted']].head(5))
```

## 8. Live Prediction Pipeline & Real-World Testing

We create a production-ready function `predict_news(headline)` that handles cleaning, feature transformation, prediction, and confidence scoring.

```python
def predict_news(headline, model=loaded_model, vectorizer=loaded_vectorizer):
    """
    Cleans an input headline and returns prediction and class probabilities.
    """
    cleaned = clean_text(headline)
    vec = vectorizer.transform([cleaned])
    pred = model.predict(vec)[0]
    
    if hasattr(model, 'predict_proba'):
        proba = model.predict_proba(vec)[0]
        fake_prob = proba[0]
        real_prob = proba[1]
    else:
        fake_prob = 1.0 if pred == 0 else 0.0
        real_prob = 1.0 if pred == 1 else 0.0

    label = "Real News" if pred == 1 else "Fake News"
    confidence = max(fake_prob, real_prob) * 100

    return {
        'headline': headline,
        'cleaned': cleaned,
        'prediction': label,
        'confidence': f"{confidence:.1f}%",
        'real_prob': f"{real_prob:.3f}",
        'fake_prob': f"{fake_prob:.3f}"
    }

# Test custom real-world headlines
sample_headlines = [
    "NASA's James Webb Telescope discovers new evidence of water on distant exoplanet",
    "SHOCKING: Secret cure for aging discovered by unknown monk, governments trying to ban it!",
    "Federal Reserve holds interest rates steady amid moderating inflation",
    "Celebrity claims drinking lemon water every morning cured all chronic diseases in 24 hours"
]

print("--- Testing Live Predictions ---\n")
for h in sample_headlines:
    res = predict_news(h)
    print(f"Headline   : {res['headline']}")
    print(f"Prediction : {res['prediction']} (Confidence: {res['confidence']})")
    print(f"Probabilities - Real: {res['real_prob']} | Fake: {res['fake_prob']}")
    print("-" * 75)
```

```python
# Random test samples from test dataset to compare Actual vs Predicted
sample_test = X_test.assign(actual=y_test).sample(10, random_state=42)

print("--- 10 Sample Predictions from Test Dataset ---\n")
for _, row in sample_test.iterrows():
    res = predict_news(row['title'])
    actual_label = "Real News" if row['actual'] == 1 else "Fake News"
    status = "CORRECT" if res['prediction'] == actual_label else "INCORRECT"
    print(f"[{status}] Actual: {actual_label} | Predicted: {res['prediction']} ({res['confidence']})")
    print(f"Headline: {row['title']}")
    print("-" * 75)
```
